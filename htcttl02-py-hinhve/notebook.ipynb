## 4. Xem các hình vừa tạo

```python
import glob
from IPython.display import Image, display
imgs = [p for p in glob.glob(PKG + "/**/*", recursive=True)
        if p.lower().endswith((".png", ".jpg", ".jpeg")) and os.path.getmtime(p) >= T0]
for p in imgs:
    print(os.path.relpath(p, PKG)); display(Image(p))
```

## 5. Tải KẾT QUẢ về máy
File zip chỉ chứa **kết quả** tạo ra trong phiên này (bảng csv/excel và hình ảnh), không chứa dữ liệu đầu vào hay mã nguồn.

```python
import zipfile
from google.colab import files
excl = {"sem_block_proposed.yaml", "sem_block_results.yaml"}
res = []
for root, _, names in os.walk(PKG):
    for nm in names:
        fp = os.path.join(root, nm); rel = os.path.relpath(fp, PKG)
        if (nm.lower().endswith((".csv", ".xlsx", ".xls", ".png", ".jpg", ".jpeg"))
                and not rel.startswith("_shared") and nm not in excl and os.path.getmtime(fp) >= T0):
            res.append((fp, rel))
assert res, "Chưa có file kết quả nào để nén."
with zipfile.ZipFile("/content/ket_qua.zip", "w", zipfile.ZIP_DEFLATED) as z:
    for fp, rel in res: z.write(fp, rel)
print("Đã nén", len(res), "file kết quả")
files.download("/content/ket_qua.zip")
```
